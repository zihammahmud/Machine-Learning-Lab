1. Step 1 to load the "Dataset"

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('/content/synthetic_employee_data - synthetic_employee_data.csv')

print(df.head())
print(df.isnull().sum())
```

2. Step 2 Handling the missing values

```python
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns
categorical_columns = df.select_dtypes(include=['object']).columns

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

for column in categorical_columns:
    if df[column].notnull().any():
        df[column] = df[column].fillna(df[column].mode()[0])

print(df.isnull().sum())
```

3. Step 3 Encoding categoricaal variables

```python
df = pd.get_dummies(df, columns=['Department', 'Education_Level', 'Location'], dtype=int)

print(df.head())
```

4. Step 4 Scaling numerical features

```python
numerical_columns = ['Age', 'Salary', 'Years_at_Company', 'Number_of_Projects', 'Performance_Score']

scaler = StandardScaler()

df[numerical_columns] = scaler.fit_transform(df[numerical_columns])

print(df.head())
```
